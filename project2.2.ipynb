```python
from datetime import datetime

def convert_dob(date_string):
    if pd.isna(date_string):
        return pd.NaT
    
    date_string = str(date_string)
    day, month, year = date_string.split("-")
    year = int(year)
    
    # Assume 00-26 means 2000-2026
    # and 27-99 means 1927-1999
    if year <= 26:
        full_year = 2000 + year
    else:
        full_year = 1900 + year
    
    return pd.Timestamp(
        year=full_year,
        month=datetime.strptime(month, "%b").month,
        day=int(day)
    )
```

```python
# Numerical features
numeric_features = [
    "amount_usd",
    "annual_rate_pct",
    "term_months",
    "num_dependents",
    "months_at_employer",
    "monthly_income_usd",
    "existing_obligations",
    "age_at_approval"
]

# Categorical features
categorical_features = [
    "product_code",
    "payment_frequency",
    "loan_purpose",
    "client_gender",
    "marital_status",
    "employment_sector",
    "collateral_type",
    "disbursement_channel",
    "province"
]

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
##processing pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.pipeline import Pipeline
```

```python
# Preprocessing for numerical variables
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Preprocessing for categorical variables
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])
```

```python
logistic_model.fit(X_train, y_train)
```

```python
print("Logistic Regression model trained successfully!")
```

```python
##making predictions
y_pred = logistic_model.predict(X_test)
```

```python
import pandas as pd
import joblib

# Load the saved model
model = joblib.load("loan_default_xgboost_model.pkl")

# Our selected classification threshold
THRESHOLD = 0.20


def predict_loan_default(applicant_data):
    
    # Convert applicant information into a DataFrame
    applicant_df = pd.DataFrame([applicant_data])
    
    # Get probability of default
    probability = model.predict_proba(applicant_df)[0, 1]
    
    # Apply our selected threshold
    prediction = int(probability >= THRESHOLD)
    
    # Determine risk level
    if prediction == 1:
        risk = "HIGH DEFAULT RISK"
        decision = "Likely to Default"
    else:
        risk = "LOW DEFAULT RISK"
        decision = "Likely to Repay"
    
    return {
        "default_probability": probability,
        "default_percentage": probability * 100,
        "prediction": prediction,
        "risk": risk,
        "decision": decision
    }
```
